# Sliding Window Patterns for Coding Interviews

Sliding window turns many O(n²) substring/subarray problems into O(n). This notebook walks through the two variants you will meet most often in interviews: **fixed-size** and **variable-size** windows, with runnable examples and the questions to ask your interviewer first.

## 1. Fixed-size window: maximum sum of k consecutive elements

Keep a running sum; when the window slides, add the new element and subtract the one leaving. Each element is touched twice, so the cost is O(n) time and O(1) space.

```python
def max_sum_k(nums, k):
    if k <= 0 or k > len(nums):
        raise ValueError('k must be between 1 and len(nums)')
    window = sum(nums[:k])
    best = window
    for i in range(k, len(nums)):
        window += nums[i] - nums[i - k]
        best = max(best, window)
    return best

print(max_sum_k([2, 1, 5, 1, 3, 2], 3))  # 9 -> [5, 1, 3]
```

## 2. Variable-size window: longest substring without repeating characters

Grow the right edge every step; shrink the left edge only while the window breaks the rule. Storing the last index of each character lets the left edge jump instead of crawl.

```python
def longest_unique(s):
    last = {}
    left = best = 0
    for right, ch in enumerate(s):
        if ch in last and last[ch] >= left:
            left = last[ch] + 1
        last[ch] = right
        best = max(best, right - left + 1)
    return best

for t in ['abcabcbb', 'bbbbb', 'pwwkew', '']:
    print(repr(t), longest_unique(t))  # 3, 1, 3, 0
```

## 3. Variable-size window with a target: minimum-length subarray with sum ≥ target

Works only when all numbers are **positive** — ask about negatives before you commit, because with negatives you need prefix sums plus a monotonic deque instead.

```python
def min_len_at_least(nums, target):
    left = total = 0
    best = float('inf')
    for right, x in enumerate(nums):
        total += x
        while total >= target:
            best = min(best, right - left + 1)
            total -= nums[left]
            left += 1
    return 0 if best == float('inf') else best

print(min_len_at_least([2, 3, 1, 2, 4, 3], 7))  # 2 -> [4, 3]
```

## How to recognise a sliding-window problem

- The input is a sequence (array or string) and the answer is about a **contiguous** part of it.
- The words *longest*, *shortest*, *at most k*, *exactly k distinct* or *every window of size k* appear.
- A brute-force solution re-scans overlapping ranges.

**Questions to ask before coding:** Can values be negative? Is the input empty-able? Unicode or ASCII only? What should be returned when no window qualifies?

## Talking through it in the interview

State the invariant out loud ("the window never contains a duplicate"), say when the left pointer moves, and finish with the complexity. Interviewers grade the reasoning as much as the code.
